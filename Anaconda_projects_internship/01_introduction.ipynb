```python
import json
```

```python
#lets write the function to load the data
def load_data(filename):
    with open(filename,"r") as f:
        data = json.load(f)
    return data 
```

```python
data = load_data("data.json")
```

```python
#wRITE A FUNCTON TO DISPLAY USERS AND THERE CONNECTION
def display_users(data):
    print("users and their connecxtions\n")
    for user in data['users']:
      print(f"ID{user['id']}:-{user ['name']} is friends with {user ['friends']} and liked pages are {user ['liked_pages']}")
    print("Pages Information")
    for page in data['pages']:
          print(f"{page['id']} : {page ['name']}")

display_users(data)
```
